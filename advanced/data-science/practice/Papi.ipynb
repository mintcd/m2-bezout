```python
import numpy as np
from pypapi import events, papi_high

def test_papi_flops():
    # 1. Define the workload (100x100 matrix multiplication)
    matrix_size = 100
    a = np.random.rand(matrix_size, matrix_size)
    b = np.random.rand(matrix_size, matrix_size)

    # 2. Start the PAPI counter for Double Precision FLOPs
    # Note: Depending on your CPU architecture, you might need to use 
    # events.PAPI_FP_OPS (all floating point) instead of PAPI_DP_OPS.
    papi_high.start_counters([events.PAPI_DP_OPS])

    # 3. Execute the function to measure
    c = np.dot(a, b)

    # 4. Stop the counters and retrieve the result
    results = papi_high.stop_counters()
    
    # Theoretical FLOPs for nxn matrix multiplication is roughly 2 * n^3
    theoretical_flops = 2 * (matrix_size ** 3)
    measured_flops = results[0]

    print(f"Matrix size: {matrix_size}x{matrix_size}")
    print(f"Theoretical FLOPs: {theoretical_flops}")
    print(f"Measured FLOPs:    {measured_flops}")

if __name__ == "__main__":
    test_papi_flops()
```
